### Working with dates

1. [Date functions](https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/functions.html#date-and-timestamp-functions)


### 1. Requirement
You are given a dataframe as below

```python
from pyspark.sql import functions as F

data_list = [(2022, 5, 18) , 
             (9999, 12, 31), 
             (-9999, 1, 1), 
             (10000, 1, 1), 
             (-10000, 1, 1), 
             (193, 5, 25),   
             (99, 5, 25),   
             (1000, 2, 29)]
```

```python
df=(
        spark.createDataFrame(data_list).toDF('year','month','day')
            .withColumn('my_date',F.concat_ws('-','year','month','day'))
            .drop('year','month','day')
)
df.display()
```

### 2. Convert strings to date
1. Spark validates the date against the Proleptic Gregorian calendar.
2. The negative years are BC, and the positive values are AD in Gregorian calander.
3. Valid dates are taken, and invalid dates throw an exception or taken as null

```python
df_cleaned=df.withColumn('valid_date',F.try_to_date(F.col('my_date'),'y-M-d'))
df_cleaned.display()
```

###3. Add, Subtract days and months to date

```python
df_cleaned=df_cleaned.withColumns({
                'add_date' : F.date_add(F.col('valid_date'),3),
                'sub_date' : F.date_sub(F.col('valid_date'),4),
                'add_months' : F.add_months(F.col('valid_date'),3),
                'sub_months' : F.add_months(F.col('valid_date'),-4)
})
```

###4. Current date, date difference, and interval

```python
df_cleaned=df_cleaned.withColumns({
                        'current_date' : F.current_date(),
                        'date_difference' : F.date_diff(F.col('add_date'), F.col('sub_date')),
                        'interval_col' : F.col('add_date') - F.col('sub_date'),
                        'date_diff': F.datediff(F.col('add_months'),F.col('sub_months'))
}).display()
```

###5. Format date

```python
df_cleaned.withColumn('fmt_date', F.date_format(F.col("valid_date"), 'yyyy-MMMM-dd EEEE HH')).display()
```
