# 1. Create a dataframe as below

   ```
     +---+--------+---+------+
     | id|    name|age|salary|
     +---+--------+---+------+
     |100|Prashant| 45| 45000|
     |101|   Tarun| 36| 33000|
     |102|   David| 48| 28000|
    +---+--------+---+------+
 ```

````python
from pyspark.sql.types import  StructType, StructField, IntegerType, ShortType, StringType, DoubleType
from pyspark.sql.functions import col
````

````python
schema= StructType([
                    StructField("id", IntegerType(), True),
                    StructField("name", StringType(), metadata={'description':'name'}),
                    StructField('age', ShortType(), True),
                    StructField('salary', DoubleType(), True)
])

df_list=[
                    (100,   'Prashant', 45, 45000),
                    (101,     'Tarun',   36,   33000), 
                    (102,     'David',   48,   28000)
]

raw_df=spark.createDataFrame(data=df_list,schema=schema)
````

# 2. Add following columns to your Dataframe
###     1. increment: 10% of the salary up to 3000 maximum increment
###     2. revised_salary: salary + increment


````python
transformed_df=raw_df.selectExpr(
                                    'id',
                                    'name',
                                    'age',
                                    'salary',
                                    'CASE WHEN (salary*10)/100<=3000 THEN (salary*10)/100 WHEN (salary*10)/100>3000  THEN 3000 END AS increment'
                                ).withColumn('revised_salary', (col('salary')+col("increment")))

transformed_df.show()                           
````

# 3. Add following columns to your Dataframe

## Replace the following column in your dataframe

## salary: current salary + increment

````python
replace_col_df=(
                raw_df.selectExpr(
                                    'id',
                                    'name',
                                    'age',
                                    'salary',
                                    'CASE WHEN (salary*10)/100<3000 THEN salary+(salary*10)/100 WHEN (salary*10)/100>=3000  THEN salary+3000 END AS revised_salary'
                                )
                                .withColumn('salary',col('revised_salary'))
                                .drop(col('revised_salary'))
                                      
            )


replace_col_df.show()
````

#4. Add a batch number (uuid) column to your dataframe

````python
import uuid
from pyspark.sql import functions as F

batch_id=str(uuid.uuid4())
uuid_batch_df=replace_col_df.withColumn('batch_id',F.lit(batch_id))

uuid_batch_df.display()
````

#4. Rename the dataframe colums as listed below
## 1. increment - annual_increment
## 2. salary - incremented_salary

````python
renamed_df=transformed_df.withColumnsRenamed({
                                            'increment' : 'annual_increment', 
                                            'revised_salary' : 'incremented_salary'
                                            })
renamed_df.display()
````
