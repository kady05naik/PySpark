### 1. Create a data frame to demo some scenarios

```python
from pyspark.sql.types import IntegerType, StringType, IntegerType, StructField, StructType

person_list = [(100, "Prashant", 30),
             (101, "David", None),
             (102, "Sushant",None),
             (103, "Abdul", 45),
             (104, "Shruti", 28)]

schema=StructType([
    StructField('id',IntegerType()),
    StructField('name', StringType()),
    StructField('age', IntegerType())
])

df=spark.createDataFrame(person_list, schema)
```
