## 1. Read data from flight_time table

````python
df=spark.read.table('dev.spark_db.flight_time')



(
    df.selectExpr(
                    "OP_CARRIER",
                    "ARR_TIME-CRS_ARR_TIME AS delay_time"
                )
                .where("(ORIGIN='AUS' and DEST='ORD') and (FL_DATE='2000-01-16')")
                .orderBy('delay_time',ascending=False)
                .limit(3)
                .display()
)
````

### 2. Find the top 3 most delayed flights on 2000-01-16 from AUS to ORD
 
  Collect the results into a list and display the output as the following.
  ```
      AA flight delayed by 5.0 minutes
      AA flight delayed by 2.0 minutes
      UA flight delayed by 2.0 minutes
  ```

````python

lst=spark.sql('''SELECT 
    OP_CARRIER, 
    ARR_TIME-CRS_ARR_TIME AS delay_time
FROM dev.spark_db.flight_time
WHERE  FL_DATE= '2000-01-16'
AND ORIGIN='AUS' AND DEST='ORD'
ORDER BY ARR_TIME-CRS_ARR_TIME DESC
LIMIT 3''').collect()
````

````python
lst
````

````python
result_dict=[row.asDict() for row in lst]
````

````python
result_dict
````

````python
for i in result_dict:
    print(i['OP_CARRIER'], 'flight delayed by', i['delay_time'].total_seconds()/60, 'minutes')
````

### 2. Find the third most delayed flights on 2000-01-16 from AUS to ORD

````python
(
    df.selectExpr(
                    "OP_CARRIER",
                    "ARR_TIME-CRS_ARR_TIME AS delay_time"
                ).where("ORIGIN='AUS' and DEST='ORD' and FL_DATE='2000-01-16'")
                .orderBy('delay_time',ascending=False)
                .limit(3)
                .offset(2)
                .display()
)

````
