```python
cleaned1_df=(
                raw_df.select(
                                col('id').cast(StringType()).alias('transaction_id'),
                                col('name').alias('customer_name'),
                                coalesce(col('dop').try_cast(DateType()),to_date(col('dop'),'dd-MM-yyyy')).alias('date_of_purchase'),
                                col('phone').alias('customer_phone'),
                                col('amount').try_cast(LongType()).alias('purchase_amount'),
                                coalesce(col('discount').try_cast(DoubleType()),lit(0)).alias('applied_discount')
                            ).filter(
                                        (col('purchase_amount').isNotNull()) &
                                        (col('purchase_amount').between(1, 200000))
                                    )
)
```

```python
# No need to import col, lit, or individual data types
cleaned_df = raw_df.selectExpr(
    "CAST(id AS STRING) as transaction_id",
    "name as customer_name",
    "coalesce(try_cast(dop AS DATE), to_date(dop, 'dd-MM-yyyy')) as date_of_purchase",
    "phone as customer_phone",
    "try_cast(amount AS LONG) as purchase_amount",
    "coalesce(try_cast(discount AS DOUBLE), 0) as applied_discount"
).filter(
    "purchase_amount IS NOT NULL AND purchase_amount BETWEEN 1 AND 200000"
)
```
