####1. Requirement
 You are given a dataframe below
 ```
   df = spark.createDataFrame([("25","07","1982")]).toDF("day", "month", "year")
 ```
 Create a date from day, month, and year.

````python
df = spark.createDataFrame([("25","07","1982")]).toDF("day", "month", "year")
df.display()
````

````python
from pyspark.sql import functions as F
from pyspark.sql.types import DateType

transe_date=df.withColumn('my_date',F.concat_ws("-","day","month", "year"))
transe_date.display()
final_date_col=transe_date.withColumn('my_date', F.coalesce(F.col('my_date').try_cast("date"),F.to_date(F.col('my_date'),"dd-MM-yyyy"))).display()
````

##2. Requirement
 You are given a dataframe below
 ```
   df = (spark.createDataFrame([("Sanjay", "Kalra", 25, "July", 1982, 19408.98)])
           .toDF("fist_name", "last_name", "day", "month", "year", "salary"))
 ```
 Create the following output
 ```
   +-----------------------------------------------+----------+
   |fun_text                                       |salary    |
   +-----------------------------------------------+----------+
   |Mr. Sanjay Kalra was born on 25th July of 1982.|$19,408.98|
   +-----------------------------------------------+----------+

````python
df = (spark.createDataFrame([("Sanjay", "Kalra", 25, "July", 1982, 28919408.98)])
          .toDF("first_name", "last_name", "day", "month", "year", "salary"))
````

````python
salary_fmt=F.format_number('salary','$,###.##') #$28,919,408.98
fun_text=F.format_string("Mr. %s %s was born on %dth %s of %d.", df.first_name, df.last_name, df.day, df.month, df.year)
(
    df.withColumns({'salary':salary_fmt,
                'fun_text': fun_text })
                .drop(df.first_name, df.last_name, df.day, df.month, df.year)
                .display()
)  
````

## 3. Requirement
 You are given a dataframe below
 ```
   df = spark.createDataFrame([("Benga`li Market", "110001"),("Adu~godi", "560030")]) \
           .toDF("address", "pin")
 ```
 Write code to fix the data problem in the address field


````python
df = spark.createDataFrame([("Benga`li Market", "110001"),("Adu~godi", "560030")]) \
          .toDF("address", "pin")
````

````python
df.withColumn('address', F.translate(F.col("address"),"`~","")).display()
````

````python
df.withColumn('address', F.regexp_replace(F.col("address"), "[`~]", "")).display()
````
