### Create a test Dataframe as shown below
 ```
   +---+---------+-----------+--------+
   | id|   source|destination|distance|
   +---+---------+-----------+--------+
   |101|   Mumbai|        Goa|     587|
   |102|   Mumbai|  Bangalore|     985|
   |102|   Mumbai|  Bangalore|     985|
   |103|    Delhi|    Chennai|    2208|
   |104|    Delhi|    Chennai|    2208|
   |105|Bangalore|    Kolkata|    1868|
   |105|Bangalore|    Kolkata|    1865|
   +---+---------+-----------+--------+
 ```

````python
from pyspark.sql.types import StructType, StringType, LongType, IntegerType, StructField

schema=StructType([
    StructField('id', IntegerType()),
    StructField('source', StringType()),
    StructField('destination', StringType()),
    StructField('distance', LongType())
])
    
data=[(101, "Mumbai", "Goa", 587),
             (102, "Mumbai", "Bangalore", 985),
             (102, "Mumbai", "Bangalore", 985),
             (103, "Delhi", "Chennai", 2208),
             (104, "Delhi", "Chennai", 2208),
             (105, "Bangalore", "Kolkata", 1868),
             (105, "Bangalore", "Kolkata", 1865)
             ]


df=spark.createDataFrame(data,schema)
df.display()
````

### 2. Select records that satisfy the follwing criteria
 1. Source is Mumbai and destination is Bangalore

### Show the following approaches
 1. Using a single filter
 2. Using two filters
 3. Using a col function expression
 4. Using a dataframe qualifier expression

````python
# 1. Using a single filter

df.filter("source == 'Mumbai' and  destination== 'Bangalore'").show()
````

````python
# 2. Using two filters

df.filter("source=='Mumbai'").filter("destination=='Bangalore'").show()
````

````python
# 3. Using a col function expression

from pyspark.sql import functions as F

df.filter(F.col('source')=='Mumbai').filter(F.col('destination')=='Bangalore').show()
df.filter((F.col('source')=='Mumbai') & (F.col('destination')=='Bangalore')).show()
````

````python
# 4. Using a dataframe qualifier expression

df.filter((df.source=='Mumbai') & (df.destination=='Bangalore')).show()
````
