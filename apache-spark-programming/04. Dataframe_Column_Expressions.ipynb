#Requirement
#Read data from flight_time and transform as below
## 1. Rename fl_date to dep_date
## 2. Compute arr_date
## 3. Following fields to represent full timestamp
###      1. crs_dep_time
###     2. dep_time
###      3. crs_arr_time
###      4. arr_time

```python
df=spark.sql('SELECT * FROM dev.spark_db.flight_time')
```

```python
trans1_df=(
    df.selectExpr(
                    'FL_DATE AS dep_date',
                    'to_date(dep_date+DEP_TIME+WHEELS_ON+TAXI_IN) AS arr_date',
                    'OP_CARRIER',
                    'OP_CARRIER_FL_NUM',
                    'ORIGIN',
                    'ORIGIN_CITY_NAME',
                    'DEST',
                    'DEST_CITY_NAME',
                    'CAST(dep_date+CRS_DEP_TIME AS TIMESTAMP) CRS_DEP_TIME',
                    'CAST(dep_date+DEP_TIME AS TIMESTAMP) DEP_TIME',
                    'WHEELS_ON',
                    'TAXI_IN',
                    'CAST(arr_date+CRS_ARR_TIME AS TIMESTAMP) CRS_ARR_TIME',
                    'CAST(arr_date+ARR_TIME AS TIMESTAMP)ARR_TIME',
                    'CANCELLED',
                    'DISTANCE'
    )
)
trans1_df.display()
```
